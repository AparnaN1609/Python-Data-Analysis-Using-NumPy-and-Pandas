## Data Analysis Using NumPy and Pandas

### 1: NumPy-Array Operations

```python
import numpy as np
```

#### Create a NumPy array for Week 1 temperatures

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

#### Inspect the array

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Size:", temperatures_w1.size)
```

#### Convert Celsius temperatures into Fahrenheit

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

#### Calculate maximum, minimum and average temperatures

```python
print("Maximum Temperature:", np.max(temperatures_w1))
```

```python
print("Minimum Temperature:", np.min(temperatures_w1))
```

```python
print("Average Temperature:", np.mean(temperatures_w1))
```

#### Perform slicing operations

First 3 days:

```python
print("First 3 Days:")
print(temperatures_w1[:3])
```

Weekend temperatures (last 2 days):

```python
print("Weekend Temperatures:")
print(temperatures_w1[-2:])
```

Middle 3 days (Days 3, 4 and 5):

```python
print("Middle 3 Days:")
print(temperatures_w1[2:5])
```

#### Create a 2D NumPy array for two weeks

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Two Weeks Temperature Data:")
print(temperatures)
```

#### Inspect the 2D array

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Size:", temperatures.size)
```

#### Extract each week's temperatures

```python
print("Week 1 Temperatures:")
print(temperatures[0])
```

```python
print("Week 2 Temperatures:")
print(temperatures[1])
```

#### Extract weekend temperatures for both weeks

```python
print("Week 1 Weekend Temperatures:")
print(temperatures[0, -2:])
```

```python
print("Week 2 Weekend Temperatures:")
print(temperatures[1, -2:])
```

```python
print("Weekend Temperatures of Both Weeks:")
print(temperatures[:, -2:])
```

### 2: Pandas – Series Operations

```python
import pandas as pd
```

#### Create a Pandas Series

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

#### Access the first rank using integer position

```python
print("First Rank Mark:")
print(marks.iloc[0])
```

#### Access the top 3 ranks using .loc

```python
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

#### Access the third rank using .iloc

```python
print("Third Rank Mark:")
print(marks.iloc[2])
```

#### Filter marks greater than 90

```python
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

#### Update the first rank mark to 100

```python
marks.loc['Rank1'] = 100

print("Updated Marks:")
print(marks)
```

#### Remove Rank5

```python
marks = marks.drop('Rank5')

print("Marks After Removing Rank5:")
print(marks)
```

#### Calculate CGPA by dividing marks by 10

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

### Pandas – DataFrame Operations

#### Create the DataFrame 

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("Transactions DataFrame:")
print(transactions)
```

#### Display the first five rows

```python
print("First Five Rows:")
print(transactions.head())
```

#### Display the last five rows

```python
print("Last Five Rows:")
print(transactions.tail())
```

#### Check the shape of the DataFrame 

```python
print("Shape:")
print(transactions.shape)
```

#### Display column names

```python
print("Column Names:")
print(transactions.columns)
```

#### Check data types

```python
print("Data Types:")
print(transactions.dtypes)
```

#### Display DataFrame information

```python
transactions.info()
```

#### Select ProductCategory and Amount columns

```python
print("Selected Columns:")
print(transactions[['ProductCategory', 'Amount']])
```

#### Select the last three columns using iloc

```python
print("Last Three Columns:")
print(transactions.iloc[:, -3:])
```

#### Select the last three columns using loc

```python
print("Last Three Columns Using loc:")
print(transactions.loc[:, ['ProductCategory', 'Region', 'Amount']])
```

#### Filter transactions from the North region with an amount greater than 200

```python
north_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print("North Region Transactions with Amount Greater Than 200:")
print(north_transactions)
```

#### Count the number of transactions in each product category

```python
print("Product Category Counts:")
print(transactions['ProductCategory'].value_counts())
```

#### Find unique regions

```python
print("Unique Regions:")
print(transactions['Region'].unique())
```

#### Calculate average transaction amount for each region

```python
print("Average Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

#### Update the amount for TransactionID 102 to 165

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("Updated Transactions:")
print(transactions)
```

#### Add a Discount column with 10% of the Amount

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print("DataFrame After Adding Discount:")
print(transactions)
```

#### Remove TransactionID 109

```python
transactions = transactions[transactions['TransactionID'] != 109]

print("DataFrame After Removing TransactionID 109:")
print(transactions)
```

#### Delete the Discount column

```python
transactions = transactions.drop('Discount', axis=1)

print("Final Transactions DataFrame:")
print(transactions)
```
